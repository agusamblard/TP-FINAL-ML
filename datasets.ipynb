```python
import pandas as pd 
from preprocessing.data_cleanse import limpiar_dataset
from preprocessing.hmv import hmv_dataset
from preprocessing.df_to_numeric import df_to_numeric
from sklearn.model_selection import train_test_split

```

```python
df = pd.read_csv('data/raw/pf_suvs_i302_1s2025.csv')
train, test = train_test_split(df, test_size=0.2, random_state=42, shuffle=True)
train, val = train_test_split(train, test_size=0.2, random_state=42, shuffle=True)

train.to_csv('data/raw/train.csv', index=False)
val.to_csv('data/raw/val.csv', index=False)
test.to_csv('data/raw/test.csv', index=False)
```

```python
train = limpiar_dataset(train)
train = hmv_dataset(train, train)
train.to_csv('data/processed/train_processed.csv', index=False)
val = limpiar_dataset(val)
val = hmv_dataset(train, val)
val.to_csv('data/processed/val_processed.csv', index=False)
test = limpiar_dataset(test)
test = hmv_dataset(train, test)
test.to_csv('data/processed/test_processed.csv', index=False)


```
