### ¿Cuál es el efecto de los años de uso, el kilometraje y si tiene camara de retroceso sobre el precio de un auto?

```python
# ¿Cuál es el efecto de los años de uso, el kilometraje y si tiene camara de retroceso sobre el precio de un auto?

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# Cargar los datos
raw_data = pd.read_csv("./data/raw/pf_suvs_i302_1s2025.csv")

from preprocessing.data_cleanse import limpiar_dataset
from preprocessing.hmv import hmv_dataset
from preprocessing.df_to_numeric import df_to_numeric

data = limpiar_dataset(raw_data)
data = hmv_dataset(data, data, precio=True)
data = df_to_numeric(data, data)


```

```python
# print correlation matrix de las variables [Precio, Año, Kilometraje, Camara de retroceso, HP, is_manual]
correlation_matrix = data[['Precio', 'Año', 'Kilómetros', 'tiene_camara', 'HP', 'is_manual']].corr()
plt.figure(figsize=(10, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matriz de Correlación')
plt.show()
```
